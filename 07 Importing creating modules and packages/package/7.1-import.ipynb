Importing Modules in Python: Modules and Packages

In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(20))
print(sqrt(50))
print(pi)
```

```python
from math import *
sin(30)
```

```python

import numpy as np
np.array([1,2,3,4])
```

```python
from package.math import addition
```

```python
import package.math as m
m.addition(10,20)
```

```python
import random
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```

```python
import os
print(os.getcwd())
```

```python
os.mkdir('test_dir')
```

```python
import shutil
shutil.copyfile('source.txt', 'destination.txt')
```

```python
dir(shutil)
```

```python
## Data serialization

import json
data = {'name': 'Mhk','age': 20}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## Datetime
from datetime import datetime,timedelta
now = datetime.now()
print(now)
yesterday = now-timedelta(days=1)
print(yesterday)
```

```python
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## regular expression

import re 
pattern = r'\d+'
text = 'There are 123 apples 456'
match = re.search(pattern, text)
print(match.group())
```
