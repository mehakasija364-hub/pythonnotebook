```python
def max_element(lst):
    if not lst:
        return None
    max_val=lst[0]
    for num in lst:
        if num>max_val:
            max_val=num
    return max_val 
print(max_element([3,5,2,9,6]))
print(max_element([-1,-2,-3,-4]))
print(max_element([]))
```

```python
def sum_element(lst):
    total=0
    for num in lst:
        total+=num
    return total
print(sum_element([1,2,3,4,5]))
```

```python
def is_palindrome(lst):
    left,right=0,len(lst)-1
    while left<right:
        if lst[left]!=lst[right]:
            return False
        left+=1
        right-=1
    return True
print(is_palindrome([7,8,9,8,7]))
print(is_palindrome([1,2,3,4,5]))
```

```python
def reverse_list(lst):
    left,right=0,len(lst)-1
    while left<right:
        lst[left],lst[right]=lst[right],lst[left]
        left+=1
        right-=1
    return lst
print(reverse_list([1,2,3,4,5]))
```

```python
def rotate_list(arr,d):
    n=len(arr)
    result=[]
    for i in range(d,n):
        result.append(arr[i])
    for i in range(0,d):
        result.append(arr[i])
    return result
print(rotate_list([1,2,3,4,5],2))
print(rotate_list([10,20,30,40,50],3))
```

```python
def reverse(nums,start,end):
    while start<end:
        nums[start],nums[end]=nums[end],nums[start]
        start+=1
        end-=1
def rotate_list1(nums,d):
    n=len(nums)
    if n==0:
        return 0
    d=d%n
    #reverse first d element
    reverse(nums,0,d-1)
    #reverse remaining items
    reverse(nums,d,n-1)
    #reverse complete list
    reverse(nums,0,n-1)
    return nums
print(rotate_list1([1,2,3,4,5],2))
       
```
