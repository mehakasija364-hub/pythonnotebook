```python
#deep copy , shallow copy
#assignment copy ()
a=[1,2,3]
b=a
b.append(4)
print(a)
print(b)
```

```python
#deep copy
#(Deep copy creates a new object and recursively copies all nested objects inside it. Changes in the copied object do not affect the original object.)
import copy
original=[[1,2],[3,4]]
copied=copy.deepcopy(original)
copied[0][0]=100
print("Original:", original)
print("Copied", copied)
```

```python
#shallow copy
a=[1,2,3]
b=a.copy()
b.append(4)
print(a)
print(b)
#nested shallow copy

import copy
original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
shallow[0][0] = 100

print("Original:", original)
print("Shallow:", shallow)
```

Tc of list operations 

accessing by index, append, pop,update by index= O(1)

insert,at beggining, remove by value, search = O(n)

Sort = O(nlog n) 

reverse and copy = O(n)

slicing of k elements = O(k)

List vs Tuple

List= [], orderd, mutable(change)
duplicates(yes)
indexing(yes)

tuple= (), ordered, immutable, duplicates, indexing

```python
# filter filters based on conditions
numbers=[1,2,3,4,5,6]
result=list(filter(lambda x:x%2==0, numbers))
print(result)
```

```python
#map  applies a function to every element
numbers=[1,2,3,4,5,6]
result=list(map(lambda x:x**2, numbers))
print(result)
```

```python
# any all
numbers=[1,2,3,4,5,6]
print(any(x>5 for x in numbers))
print(all(x%2==0 for x in numbers))
```
