```python
def count_negative(grid):
    count=0
    n=len(grid[0])
    for row in grid:
        left,right=0,n-1
        first_negative=n
        while left<=right:
            mid=(left+right)//2
            if row[mid]<0:
                first_negative=mid
                right=mid-1
            else:
                left=mid+1
        count+=n-first_negative
    return count
grid = [
    [4, 3, 2, 1],
    [3, 2, 1, -1],
    [1, 1, -1, -2],
    [-1, -1, -2, -3]
]
for row in grid:
    print(row)
result = count_negative(grid)
print(result)

     
```

```python
def smallest_letter(words,target):
   left,right=0,len(words)-1
   ans=0
   while left<=right:
      mid=(left+right)//2
      if (ord(words[mid])>ord(target)):
         ans=mid
         right=mid-1
      else:
         left=mid+1
   return words[ans]
print(smallest_letter(['c','f','j'],'k'))
print(smallest_letter(['c','f','j'],'c'))
print(smallest_letter(['c','f','j'],'a'))
```

```python
def first_and_last_pos(nums,target):
    left,right=0,len(nums)-1
    ans=-1
    while left<=right:
        mid=(left+right)//2
        if nums[mid]==target:
            ans=mid
            right=mid-1
        elif nums[mid]<target:
            left=mid+1
        else:
            right=mid-1  
    
    left,right=0,len(nums)-1
    ans1=-1
    while left<=right:
            mid=(left+right)//2
            if nums[mid]==target:
                ans1=mid
                left=mid+1
            elif nums[mid]<target:
                left=mid+1
            else:
                right=mid-1
    return [ans,ans1]
print(first_and_last_pos([5,7,7,8,8,10],8))    
```

```python
def find_min(nums):
    left,right=0,len(nums)-1
    while left<right:
        mid=(left+right)//2
        if nums[mid]>nums[right]:
            left=mid+1
        else:
            right=mid
    return nums[left]
print(find_min([4,5,6,7,0,1,2]))
print(find_min([11,13,15,17]))
    
```

```python
def rotatedarray(nums,target):
    left,right=0,len(nums)-1
    while left<=right:
        mid=(left+right)//2
        if nums[mid]==target:
            return mid
        #left
        if nums[left]<=nums[mid]:
            if nums[left]<=target<nums[mid]:
                right=mid-1
            else:
                left=mid+1
        #right
        else:
            if nums[mid]<target<=nums[right]:
                left=mid+1
            else:
                right=mid-1
    return -1
print(rotatedarray([4,5,6,7,0,1,2],0))    

```
