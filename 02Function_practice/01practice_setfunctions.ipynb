```python
def max_lift(n, capacity):
    rounds=0
    people=n
    while(people>0):
        people=people-capacity
        rounds+=1
    return rounds

n=10
capacity=3
print(max_lift(n,capacity))
```

```python
def slope_find(slope,intercept,x):
    return slope*x+intercept
print(slope_find(2,3,4))
```

```python
def sum_all(numbers):
    return sum(numbers)
print(sum_all([1,2,3,4,5]))
```

```python
def largest_element(numbers):# O(n)
    largest=numbers[0]
    for n in numbers:
        if n>largest:
            largest=n
    return largest
print(largest_element([3,8,2,10,5]))
```

```python
def remove_duplicates(lst):
    return list(set(lst))
print(remove_duplicates([1,2,2,3,4,4,5]))
```

```python
def remove_duplicates(lst):
    result=[]
    for num in lst:
       if num not in result:
           result.append(num)
    return result
print(remove_duplicates([1,2,2,3,4,4,5]))
```

```python
def remove_duplicate(lst):# tc O(n)
    s={}
    for num in lst:
        if num in s:
            return False
            s[num]=True
    return True
print(remove_duplicates([1,2,2,3,4,4,5]))
```
