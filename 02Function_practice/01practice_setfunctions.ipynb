```python
def max_lift(n, capacity):
    rounds=0
    people=n
    while(people>0):
        people=people-capacity
        rounds+=1
    return rounds

n=10
capacity=3
print(max_lift(n,capacity))
```

```python
def largest_element(numbers):# O(n)
    largest=numbers[0]
    for n in numbers:
        if n>largest:
            largest=n
    return largest
print(largest_element([3,8,2,10,5]))
```

```python
def remove_duplicates(lst):
    result=[]
    for num in lst:
       if num not in result:
           result.append(num)
    return result
print(remove_duplicates([1,2,2,3,4,4,5]))
```

```python
def remove_duplicate(lst):# tc O(n)
    #s={}
    s=set()
    for num in lst:
        if num in s:
            return False
       # s[num]=True
        s.add(num)
    return True
print(remove_duplicate([1,2,2,3,4,4,5]))
```

```python
def reverse_list(nums):
    left=0
    right=len(nums)-1
    while left<right:
      nums[left], nums[right]=nums[right],nums[left]
      left+=1
      right-=1
    return nums
print(reverse_list([1,2,3,4,5]))
```

```python
def count_even_odd(lst):
   even_count=len([x for x in lst if x%2==0])
   odd_count=len([x for x in lst if x%2!=0])
   return (even_count,odd_count)
print(count_even_odd([1,2,3,4,5]))
```

```python
def max_consective_diff(lst):
    max_diff=0
    for i in range(len(lst)-1):
        diff=abs(lst[i]-lst[i+1])
        if diff>max_diff:
            max_diff=diff
    return max_diff
print(max_consective_diff([1,7,3,10,5]))
```

```python
def merge_lists(list1,list2):
    result=[]
    i=0
    j=0
    while i<len(list1) and j<len(list2):
        if list1[i]<list2[j]:
            result.append(list1[i])
            i+=1
        else:
            result.append(list2[j])
            j+=1
    result.extend(list1[i:])
    result.extend(list2[j:])
    return result
print(merge_lists([1,3,5],[2,4,6]))
```

```python
def rotate_list(lst,k):
    n=len(lst)
    result=[]
    for i in range(n-k,n):
        result.append(lst[i])
    for i in range(0,n-k):
        result.append(lst[i])
    return result
print(rotate_list([1,2,3,4,5],2))
```

```python
def merge_dict(keys,values):
    result={}
    for i in range(len(keys)):
       result[keys[i]]=values[i]
    return result
print(merge_dict(['a','b','c'],[1,2,3]))
```

```python
def merge_three_dict(dict1,dict2,dict3):
    result={}
    for i in (dict1,dict2,dict3):
       for key in i:
           result[key]=i[key]
    return result
dict1={'a':1,'b':2}
dict2={'c':3,'d':4}
dict3={'e':5,'f':6}
merge_three_dict(dict1,dict2,dict3)
#print(merge_dict({'a':1,'b':2},{'c':3,'d':4},{'e':5,'f':6}))
```

```python
def count_word_freq(sentence):
    result={}
    words=sentence.split()
    for word in words:
        if word in result:
          result[word]=result[word]+1
        else:
          result[word]=1
    return result
print(count_word_freq("hello world hello"))
```
