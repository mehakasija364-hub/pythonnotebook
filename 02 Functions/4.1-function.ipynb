```python
def add(a,b):
    return a+b
result=add(2,3)
print(result)
```

```python
# default parameters
def greet(name="guests"):
    print(f"Hello {name} Welcome too paradise")
greet()
```

```python
# Variable length arguments
def print_numbers(*Mehak):
    for number in Mehak:
        print(number)
print_numbers(1,2,3,4,5,6,7,8,"Mehak")        
```

```python
# positional argument
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
# keyword argument
def print_details(**kwargs):
   for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(name="Mhk",age=20,country="India")
```

```python
# Both args, and kwargs
def print_details(*args, **kwargs):
    for val in args:
        print(f"Position argument: {val}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(1,2,3,4,"Mhk", name="apple", color="Red")
```
