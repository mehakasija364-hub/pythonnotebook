Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
empty_dict={}
print(type(empty_dict))
```

```python
student={"name":"Mhk", "age":20,"grade":94}
print(student)
print(type(student))
print(student["name"])
print(student.get('grade'))
print(student.get('last_name', " Not Available"))

```

```python
fruits={"name": "strawberry", "color":"pink"}
print(fruits)
```

```python
keys=fruits.keys()
print(keys)
```

```python
values=fruits.values()
print(values)
```

```python
fruits_copy=fruits
print(fruits)
print(fruits_copy)
```

```python
fruits["name"]="cherry"
fruits["color"]="red"
print(fruits)
print(fruits_copy)
```

```python
# iterating over dictionary
for value in fruits.values():
    print(value)
```

```python
for key,value in fruits.items():
    print(f"{key}: {value}")
```

```python
# nested dictionary
students={
    "student1":{"name":"Mhk", "age":20},
    "student2":{"name":"Diksha", "age":20}
}
print(students)
```

```python
# Access
print(students["student1"]["name"])
```

```python
stationary={"pen":"ball", "pencil":"natraj"}
print(stationary)
```

```python
stationary.items()
```

```python
students.items()
```

```python
# ditionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
# practice Examples
## Use a dictrionary to count the frequncy of element in the list

numbers=[1,2,2,3,3,3,4,4,4,4]
freq={}
for number in numbers:
 if number in freq:
  freq[number]+=1
 else:
  freq[number]=1
print(freq)
```

```python
# merge 2 dictionary
dict1={"a":1, "b":2}
dict2={"b":3, "c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```

Conclusion
Dictionaries are powerful tools in Python for managing key-value pairs. They are used in a variety of real-world scenarios, such as counting word frequency, grouping data, storing configuration settings, managing phonebooks, tracking inventory, and caching results. Understanding how to leverage dictionaries effectively can greatly enhance the efficiency and readability of your code.
