Program 1: Find largest element

```python
numbers = [1, 2, 3, 4, 5]

result = [x for x in numbers if x % 2 != 0]

print(result)
```

Program 2: Find smallest element

```python
numbers = [10, 50, 20, 80, 30]

smallest = min(numbers)

print("Smallest:", smallest)
```

Program 3: Find sum of list

```python
numbers = [10, 20, 30, 40]

total = 0

for num in numbers:
    total += num

print(total)
```

Program 4: Count even and odd numbers

```python
numbers = [1, 2, 3, 4, 5, 6]

even_count = 0
odd_count = 0

for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even:", even_count)
print("Odd:", odd_count)
```

Program 5: Remove duplicates

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

unique = list(set(numbers))

print(unique)
```

```python
#Note: Set order is not guaranteed. To preserve original order:
numbers = [1, 2, 2, 3, 4, 4, 5]

unique = list(dict.fromkeys(numbers))

print(unique)
```

Program 6: Reverse a list without reverse()

```python
numbers = [1, 2, 3, 4, 5]

result = []

for num in numbers:
    result.insert(0, num)

print(result)
```

Program 7: Find second largest element

```python
numbers = [10, 50, 20, 80, 30]

unique = list(set(numbers))

unique.sort()

print("Second largest:", unique[-2])
```

Program 8: Find common elements in two lists

```python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

common = [x for x in a if x in b]

print(common)
```

Program 9: Flatten a nested list

```python
matrix = [[1, 2], [3, 4], [5, 6]]

result = []

for row in matrix:
    for num in row:
        result.append(num)

print(result)
```

Program 10: Rotate list by one position

```python
numbers = [1, 2, 3, 4, 5]

last = numbers.pop()

numbers.insert(0, last)

print(numbers)
```
