Ex 2: Orgnizing Student Grades

```python
grades=[85,92,78,90,88]
grades.append(95)
avg_marks=sum(grades)/len(grades)
print(f"Avg Grade:{avg_marks:.2f}")
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Highest Grade: {highest_grade}")
print(f"Lowest Grade: {lowest_grade}")
      
```

Ex 3 Managing an Inventory


```python
inventory = ["apples", "bananas", "oranges", "grapes"]

# Adding a new item
inventory.append("strawberries")

# Removing an item that is out of stock
inventory.remove("bananas")

# Checking if an item is in stock
item = "berry"
if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} are out of stock")

# Printing the inventory
print("Inventory List:")
for item in inventory:
    print(f"- {item}")
```

Ex 4: Collecting User Feedback

```python
feedback=["good", "VERY NICE", "not satisfied"]
# Adding new feedback
feedback.append("Not happy, bad service")
pos_feedback_count=sum(1 for comment in feedback if "good" in comment.lower() or "VERY NICE" in comment.upper())
print(f"Positive feedback count: {pos_feedback_count}")

print("User Feedback:")
for comment in feedback:
    print(f"- {comment}")
```
