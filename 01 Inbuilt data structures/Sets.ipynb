Sets

Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
## Create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
numbers={1,2,3,4,5,6}
print(numbers)
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
# Adding removing element
my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
my_set.remove(4)
print(my_set)
```

```python
my_set.discard(8)
print(my_set)
```

```python
x=my_set.pop()
print(x)
print(my_set)
```

```python
my_set.clear()
print(my_set)
```

```python
# set membership opearation
my_set={1,2,3,4,5,6,7,8}
print(3 in my_set)
print(10 in my_set)
```

```python
# Mathematical Opearions
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
union_set=set1.union(set2)
print(union_set)
```

```python
#intersection
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
intersection_set=set1.intersection(set2)
print(intersection_set)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
diff=set1.difference(set2)
print(diff)
```

```python
#symmetric diff
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
diff=set1.symmetric_difference(set2)
print(diff)
```

```python
# Set methods
set1={1,2,3,4,5}
set2={3,4,5}
## is subset
print(set1.issubset(set2))
```

```python
set1={1,2,3,4,5}
set2={3,4,5}
## is subset
print(set1.issuperset(set2))
```

```python
set1={1,2}
set2={4,5}
## is subset
print(set1.isdisjoint(set2))
```

```python
lst=[1,2,3,4]
set(lst)
```

```python
text="I am Mehak and I am 20 yrs old"
words=text.split()
print(words)
# unique(use sets)
unique_words=set(words)
print(unique_words)
print(len(unique_words))

```

Conclusion

###### Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
