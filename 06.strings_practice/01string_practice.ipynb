```python
def reverse_string(s):
    result=list(s)
    left=0
    right=len(s)-1
    while left<right:
        result[left],result[right]=result[right],result[left]
        left+=1
        right-=1
    return "".join(result)
print(reverse_string("hello"))
```

```python
def count_vowels(s):
    vowels="aeiouAEIOU"
    count=0
    for ch in s:
        if ch in vowels:
            count+=1
    return count
print(count_vowels("Hello, World!"))
```

```python
def is_palindrome(s):
    left=0
    right=len(s)-1
    while left<right:
        if not s[left].isalnum():
            left+=1
            continue
        if not s[right].isalnum():
            right-=1
            continue
        if s[left].lower()!=s[right].lower():
            return False
        left+=1
        right-=1
    return True
print(is_palindrome("A man a plan a canal Panama")) 
```

```python
def count_words(s):
    #word=s.split()
    #return len(word)
    count=0
    result=False
    for ch in s:
        if ch!=" ":
            if result==False:
                count=count+1
                result=True
        else:
          result=False   
    return count        
print(count_words("Hello, World"))
```

```python
def remove_duplicates(s):
    result=""
    for ch in s:
        if ch not in result:
            result=result+ch
    return result
print(remove_duplicates("programming"))
```

```python
def remove_duplicate1(s):
    seen=set()
    result=[]
    for ch in s:
        if ch not in seen:
            seen.add(ch)
            result.append(ch)
    return "".join(result)

print(remove_duplicate1("programming"))
```

```python
def count_consonents(s):
    vowels="aeiouAEIOU"
    count=0
    for ch in s:
        if ch.isalpha() and ch  not in vowels:
            count+=1
    return count
print(count_consonents("Hello, World!"))
```

```python
def is_anagram(s,t):
    if len(s)!=len(t):
        return False
    return sorted(s)==sorted(t)
print(is_anagram("anagram","nagaram"))
```

```python
def is_anagram(s,t):
    if len(s)!=len(t):
        return False
    freq={}
    freq1={}
    for ch in s:
        freq[ch]=freq.get(ch,0)+1
    for ch in t:
         freq1[ch]=freq1.get(ch,0)+1
    return freq==freq1
print(is_anagram("anagram","nagaram"))
    
```

```python
def is_subsequence(s,t):
    j=0
    for i in range(len(s)):
        if s[i]==t[j]:
            j+=1
    return j==len(t)
print(is_subsequence("abcde","ace"))
print(is_subsequence("abcde","aec"))
```

```python
def is_substring(s,t):
    i=0
    j=0
    while i<len(s):
        if s[i]==t[j]:
            i+=1
            j+=1
            if j==len(t):
                return True
        else:
                i=i-j+1
                j=0
    return False
print(is_substring("hello world","world"))
```
