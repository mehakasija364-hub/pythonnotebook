```python
#Right Triangle
n=5
for i in range(n):
    for j in range(i+1):
        print("*", end=" ")
    print()
```

```python
#Inverted Triangle
n=5
for i in range(n):
    for j in range(n-i):
        print("*", end=" ")
    print()
```

```python
# Number Triangle
n=5
for i in range(n):
    for j in range(i+1):
        print(j+1, end=" ")
    print()   
```

```python
# Repeated Number Pattern
n=5
for i in range(n):
    for j in range(i+1):
        print(i+1, end=" ")
    print()
```

```python
# Pyramid Pattern
n=5
for i in range(n):
    for j in range(n-i-1):
        print(" ", end=" ")
    for j in range(2*i+1):
        print("*", end=" ")
    print()    
```

```python
# Inverted Number Pattern
n=5
for i in range(n):
    for j in range(n-i):
        print(j+1, end=" ")
    print()
```

```python
# Floyd's Triangle
n=4
count=1
for i in range(n):
    for i in range(i+1): 
        print(i+1, end=" ")
        count+=1
    print()
```
