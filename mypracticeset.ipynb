```python
#square
# n=4
# for i in range(n):
#     for j in range(n):
#         print("*", end=" ")
#     print()
def rectangle(n):
    result=[]
    for i in range(n):
        result.append("*"*n)
    return result
print("\n".join(rectangle(4)))
```

```python
#Right Triangle
#n=5
#for i in range(n):
#    for j in range(i+1):
#        print("*", end=" ")
#    print()
def righttriangle(n):
    result=[]
    for i in range(1,n+1):
        result.append("*"*i)
    return result
print("\n".join(righttriangle(4)))
```

```python
#Inverted Triangle
#n=5
#for i in range(n):
#    for j in range(n-i):
#        print("*", end=" ")
#    print()
def inverted_triangle(n):
    result=[]
    for i in range(n,0,-1):
        result.append("*"*i)
    return result
print("\n".join(inverted_triangle(4)))
```

```python
def same_no(n):
    result=[]
    for i in range(1,n+1):
        row=" ".join([str(i)]*i)
        result.append(row)
    return result
print("\n".join(same_no(4)))
```

```python
# Pyramid Pattern
# n=5
# for i in range(n):
#     for j in range(n-i-1):
#         print(" ", end=" ")
#     for j in range(2*i+1):
#         print("*", end=" ")
#     print()    
def pyramid(n):
    result=[]
    for i in range(0,n):
        spaces=" "*(n-i-1)
        stars="*"*(2*i+1)
        result.append(spaces+stars)
    return result
print("\n".join(pyramid(5)))
```

```python
# Inverted Number Pattern
n=5
for i in range(n):
    for j in range(n-i):
        print(j+1, end=" ")
    print()
```

```python
def alphabet(n):
    result=[]
    for i in range(1,n+1):
        row=" ".join(chr(65+j) for j in range(i))
        result.append(row)
    return result
print("\n".join(alphabet(5)))
```

```python
def hollow_rectangle(rows,cols):
      result=[]
      for i in range(rows):
            row=[]
            for j in range(cols):
                  if i==0 or i==rows-1 or j==0 or j==cols-1:
                        row.append("*")
                  else:
                        row.append(" ")
            result.append(" ".join(row))
      return result
print("\n".join(hollow_rectangle(4,5)))
```

```python
# Diamond Pattern
n=4
for i in range(n):
    for j in range(n-i-1):
        print(" ", end=" ")
    for j in range(2*i+1):
        print("*", end=" ")
    print()
for i in range(n):
    for k in range(i+1):
        print(" ", end=" ")
    for k in range(2*(n-i-2)+1):
        print("*", end=" ")
    print()  
```
