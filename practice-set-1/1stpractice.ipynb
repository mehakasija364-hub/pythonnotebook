```python
def square(n):
    result = []

    for i in range(n):
        result.append("*"*n)

    return result

print(square(3))
```

```python
def hollow_square(n):
    result=[]
    for i in range(n):
        if i==0 or i==n-1:
            result.append('*'*n)
        else:
            row='*'+' '*(n-2) + '*'
            result.append(row)

    return result
print(hollow_square(5))
```

```python
def pyramid_pattern(n):
    result=[]
    for i in range(n):
        result.append(" "*(i)+"*"*(2*(n-i)-1))
    for i in range(1,n):
        result.append(" "*(n-i-1)+"*"*(2*i+1))
    return result
l1=pyramid_pattern(3)
for i in l1:
    print(i)

```
