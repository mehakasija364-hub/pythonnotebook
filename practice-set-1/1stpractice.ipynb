```python
def square(n):
    result = []

    for i in range(n):
        result.append("*"*n)

    return result

print(square(3))
```

```python
def hollow_square(n):
    result=[]
    for i in range(n):
        if i==0 or i==n-1:
            result.append('*'*n)
        else:
            row='*'+' '*(n-2) + '*'
            result.append(row)

    return result
print(hollow_square(5))
```

```python
def rectangle(n,m):
    result=[]
    for i in range(n):
        result.append('*'*m)
    return result

rectangle(4,5)
```

```python
def right_angle_triangle(n):
    result=[]
    for i in range(1,n+1):
        result.append("*"*i)
    return result

right_angle_triangle(5)
```
