```python
#linear search
def linear_search(arr,target):
    for i in range(len(arr)):
        if(arr[i]==target):
            return i
    return -1
print(linear_search([10,20,30,50,40],30))
```

```python
#Binary Search
def binary_search(arr,target):
    low=0
    high=len(arr)-1
    while low<=high:
       # mid=(low+high)//2
        mid=low+(high-low)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]<target:
            low=mid+1
        else:
            high=mid-1
    return -1
print(binary_search([10,20,30,40,50],40))
```

```python
#bubble sort
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(n-1-i):
            if(arr[j]>arr[j+1]):
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
print(bubble_sort([5,7,9,2,1]))
```

```python
#selection sort
def selection_sort(arr):
    n=len(arr)
    for i in range(n):
        min_index=i
        for j in range(i+1,n):
            if arr[j]<arr[min_index]:
               min_index=j
        arr[i],arr[min_index]=arr[min_index],arr[i]
    return arr
print(selection_sort([5,2,1,7,3]))
```

```python
#insertion sort
def insertion_sort(arr):
    n=len(arr)
    for current in range(1,n):
      currentCard=arr[current]
      correctPos=current-1
      while correctPos>=0:
         if arr[correctPos]<currentCard:
            break
         else:
            arr[correctPos+1]=arr[correctPos]
            correctPos-=1
            arr[correctPos+1]=currentCard
    return arr
print(insertion_sort([2,4,7,1,8]))
```

```python
#insertion sort
def insertion_sort(arr):
    for i in range(1,len(arr)):
        key=arr[i]
        j=i-1
        while j>=0 and arr[j]>key:
           arr[j+1]=arr[j]
           j=j-1
           arr[j+1]=key
        return arr
print(insertion_sort([5,3,2,8]))
```

```python
def is_palindrome_tuple(ans):
    return ans==ans[::-1]
print(is_palindrome_tuple((1,2,3,2,1)))
print(is_palindrome_tuple(('a','b','c','b','a')))
```

```python
def is_palindrome_tuple(ans):
    left=0
    right=len(ans)-1
    while(left<right):
        if ans[left]!=ans[right]:
           return False
        left+=1
        right-=1
    return True
print(is_palindrome_tuple((1,2,3,2,1)))
```

```python
#Merge Dictionaries
def merge_dict_add(dicts):
    result={}
    for d in dicts:
        for key in d:
            if key in result:
                result[key]+=d[key]
            else:
                result[key]=d[key]
    return result
print(merge_dict_add([{'a':1,'b':2},{'b':3,'c':4},{'c':5,'d':6}]))
```

```python
def is_subset(lst1,lst2):
    return set(lst1).issubset(set(lst2))
print(is_subset([1,2,4],[1,2,3,4,5]))
```

```python
def is_subset(lst1,lst2):
    i=0
    while i<len(lst1):
        found=False
        j=0
        while j<len(lst2):
            if lst1[i]==lst2[j]:
                found=True
                break
            j+=1
            if found==False:
                return False
            i+=1
        return True
print(is_subset([1,2,3],[1,2,3,4,5]))
```
