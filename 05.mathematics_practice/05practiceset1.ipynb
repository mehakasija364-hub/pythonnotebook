```python
def sum_even(n):
    total=0
    num=2
    for i in range(n):
        total+=num
        num+=2
    return total
print(sum_even(3))
```

```python
def is_even(n):
    return n%2==0
print(is_even(2))
```

```python
#Prime No
def is_prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
print(is_prime(7))
```

```python
def perfect_square(num):
   if num<1:
      return False
   left=1
   right=num
   while left<=right:
      mid=(left+right)//2
      sq=mid*mid
      if sq==num:
         return True
      elif sq<num:
         left=mid+1
      else:
         right=mid-1
   return False
print(perfect_square(16))

   
       
```

```python
def decimal_to_binary(n):
    if n==0:
      return "0"
    sign=""
    if n<0:
       sign="-"
       n=abs(n)
    result=0
    place=1
    while n>0:
       rem=n%2
       result+=rem*place
       place=place*10
       n=n//2
    return sign+str(result)
print(decimal_to_binary(13))
```

```python
#Binary to decimal
def binary_to_decimal(n):
    if n==0:
      return "0"
    sign=""
    if n<0:
       sign="-"
       n=abs(n)
    result=0
    place=1
    while n>0:
       digit=n%10
       result=result+digit*place
       place=place*2
       n=n//10
    return sign+str(result)
print(binary_to_decimal(110))
```
